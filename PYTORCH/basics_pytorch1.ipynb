```python
# ============================================================
# PYTORCH FUNDAMENTALS — PRACTICE CODE
# ============================================================

import torch
import numpy as np

# 1. REPRODUCIBILITY
torch.manual_seed(42)
# Used for: Reproducing random results during experiments.


# 2. CPU / GPU
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)
# Used for: Selecting GPU when available, otherwise CPU.


# 3. CREATING TENSORS
x = torch.rand(2, 3)
y = torch.rand(3, 2)

x = x.to(device)
y = y.to(device)

print("Tensor x:\n", x)
print("Shape:", x.shape)
print("Dimensions:", x.ndim)
print("Data type:", x.dtype)
print("Device:", x.device)


# 4. BASIC OPERATIONS
print("Addition:\n", x + x)
print("Multiplication:\n", x * x)
# Used for: Element-wise mathematical operations.


# 5. MATRIX MULTIPLICATION
result = torch.matmul(x, y)
# Alternatively: result = x @ y

print("Matrix multiplication:\n", result)
print("Result shape:", result.shape)
# (2, 3) @ (3, 2) produces (2, 2).


# 6. AGGREGATION
print("Mean:", x.mean())
print("Sum:", x.sum())
print("Maximum:", x.max())
print("Maximum index:", x.argmax())
# Used for: Statistics and finding the predicted class
# from a one-dimensional tensor of class scores.


# 7. INDEXING & SLICING
print("First row:", x[0])
print("First column:", x[:, 0])
# Used for: Accessing particular elements or subsets.


# 8. RESHAPING
z = torch.arange(1, 7)
reshaped = z.reshape(2, 3)

print("Original:", z)
print("Reshaped:\n", reshaped)
# Used for: Preparing tensors for model input.


# 9. UNSQUEEZE / SQUEEZE
batch = z.unsqueeze(0)
print("Added dimension:", batch.shape)  # (1, 6)

original = batch.squeeze()
print("Removed dimension:", original.shape)  # (6,)
# Used for: Adding or removing dimensions, such as batch size.


# 10. NUMPY ↔ PYTORCH
array = np.array([1, 2, 3], dtype=np.float32)

tensor = torch.from_numpy(array)
# NumPy array → PyTorch tensor

back_to_numpy = tensor.numpy()
# CPU tensor → NumPy array

print("Tensor:", tensor)
print("NumPy:", back_to_numpy)

# Note: .numpy() requires a CPU tensor that does not require
# gradients. Use tensor.detach().cpu().numpy() when needed.
```
